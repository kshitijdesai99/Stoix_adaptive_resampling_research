```python
import pandas as pd
```

```python
df = pd.read_csv("analysis.csv")
df
```

```python
df1 = df.loc[
    (df["exp_id"].isin([5,8]))
]

with pd.option_context('display.max_rows', None, 'display.max_columns', None):
    display(df1)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Create mapping for clean labels
label_mapping = {
    'False': 'Without Residual',
    'True': 'With Residual'
}

# Convert use_residual to string and apply clean labels
df_sorted = df1.copy()
df_sorted['use_residual'] = df_sorted['use_residual'].astype(str)
df_sorted['label'] = df_sorted['use_residual'].map(label_mapping)
df_sorted = df_sorted.sort_values('episode_return_mean', ascending=False)

# Create the barplot with different colors
ax = sns.barplot(x="label", y="episode_return_mean", data=df_sorted, 
                 hue="label", palette='viridis', legend=False)

# Add labels to each bar
for container in ax.containers:
    ax.bar_label(container)

# Set clean axis labels (remove x-axis label)
ax.set_xlabel('')
ax.set_ylabel('Episode Return Mean', fontweight='bold')

# Make tick labels bold
plt.setp(ax.get_xticklabels(), fontweight='bold')
plt.setp(ax.get_yticklabels(), fontweight='bold')

plt.show()
```

```python
import pandas as pd

# Create pivot table showing mean episode return by use_residual
pivot = df1.pivot_table(
    values='episode_return_mean',
    index='use_residual',
    aggfunc='mean'
).sort_values('episode_return_mean', ascending=False)

print(pivot)
```

```python
df2 = df.loc[
    (df["total_timesteps"] == 3000000) &
    (df["ess_threshold"].isin([0,0.5]))
]
df2
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a mapping dictionary
period_mapping = {
    0: 'Adaptive Resampling\n(ESS=0.5)',
    4: 'Periodic Resampling\n(period=4)', 
    33: 'No Resampling'
}

# Map the period values to new labels
df2_mapped = df2.copy()
df2_mapped['period_label'] = df2_mapped['period'].map(period_mapping)

# Sort dataframe by episode_return_mean in descending order
df2_sorted = df2_mapped.sort_values('episode_return_mean', ascending=False)

# Create the plot with correct column name and remove error bars
ax = sns.barplot(x="period_label", y="episode_return_mean", data=df2_sorted, errorbar=None,
                 hue="period_label", palette='viridis', legend=False)

# Add labels to each bar
for container in ax.containers:
    ax.bar_label(container)

# Set y-axis label and remove x-axis label
ax.set_xlabel('')
ax.set_ylabel('Episode Return Mean', fontweight='bold')

# Make tick labels bold
plt.setp(ax.get_xticklabels(), fontweight='bold')
plt.setp(ax.get_yticklabels(), fontweight='bold')

plt.show()
```

```python
import pandas as pd

# Create pivot table showing mean episode return by period
pivot = df2.pivot_table(
    values='episode_return_mean',
    index='period',
    aggfunc='mean'
).sort_values('episode_return_mean', ascending=False)

# Map period to readable labels
period_labels = {0: 'ESS=0.5', 4: 'period=4', 33: 'No Resampling'}
pivot.index = pivot.index.map(period_labels)

print(pivot)
```

```python
# Plot for experiments 8, 12, 6 - Episode Return Mean vs Time Step
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load data
df = pd.read_csv('timeseries.csv')

# Filter for experiments 8, 12, 6
exp_ids = [8, 12, 6]

# Create mapping for experiment IDs to labels
exp_labels = {
    12: 'No Resampling',
    6: 'ESS 0.5', 
    8: 'Periodic Resampling'
}

# Get final returns for each experiment and sort by descending final return
final_returns = {}
for exp_id in exp_ids:
    exp_data = df[df['exp_id'] == exp_id].copy()
    if len(exp_data) > 0:
        final_returns[exp_id] = exp_data['episode_return_mean_eval'].iloc[-1]

# Sort experiments by final return (descending)
sorted_exps = sorted(final_returns.keys(), key=lambda x: final_returns[x], reverse=True)
colors = plt.cm.tab10(np.linspace(0, 1, len(sorted_exps)))

# Create the plot
plt.figure(figsize=(12, 7))

for i, exp_id in enumerate(sorted_exps):
    exp_data = df[df['exp_id'] == exp_id].copy()
    if len(exp_data) > 0:  # Check if data exists for this experiment
        final_return = final_returns[exp_id]
        label = f'{exp_labels[exp_id]} - {final_return:.0f}'
        plt.plot(exp_data['timestep'], exp_data['episode_return_mean_eval'], 
                 'o:', color=colors[i], linewidth=2, markersize=5, 
                 label=label, alpha=0.8)

plt.xlabel('Time Step', fontsize=12, fontweight='bold')
plt.ylabel('Episode Return Mean (Eval)', fontsize=12, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend(loc='upper left', title_fontproperties={'weight': 'bold', 'size': 12}, prop={'weight': 'bold', 'size': 11})

# Add some styling
plt.style.use('seaborn-v0_8')
plt.tight_layout()
plt.show()

# Print statistics summary
print("Statistics Summary for Experiments 8, 12, 6:")
print("-" * 50)
summary_data = []

for exp_id in sorted_exps:  # Use sorted order for display
    exp_data = df[df['exp_id'] == exp_id].copy()
    if len(exp_data) > 0:
        initial_return = exp_data['episode_return_mean_eval'].iloc[0]
        final_return = exp_data['episode_return_mean_eval'].iloc[-1]
        improvement = final_return - initial_return
        
        summary_data.append({
            'Exp_ID': exp_id,
            'Label': exp_labels[exp_id],
            'Initial': initial_return,
            'Final': final_return,
            'Improvement': improvement,
            'Data_Points': len(exp_data)
        })

# Create summary DataFrame
summary_df = pd.DataFrame(summary_data)
print(summary_df.to_string(index=False))

# Find best performing experiment
best_exp = summary_df.loc[summary_df['Improvement'].idxmax()]
print(f"\nBest performing experiment: {best_exp['Label']} (ID {best_exp['Exp_ID']}) with improvement of {best_exp['Improvement']:.2f}")
```
