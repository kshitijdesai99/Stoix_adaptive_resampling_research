```python
df = pd.read_csv("analysis.csv")
df
```

```python
df1 = df.loc[
    (df["exp_id"].isin([19,20]))
]
df1
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Create mapping for clean labels
label_mapping = {
    'False': 'Without Residual',
    'True': 'With Residual'
}

# Convert use_residual to string and apply clean labels
df_sorted = df1.copy()
df_sorted['use_residual'] = df_sorted['use_residual'].astype(str)
df_sorted['label'] = df_sorted['use_residual'].map(label_mapping)
df_sorted = df_sorted.sort_values('episode_return_mean', ascending=False)

# Create the barplot with different colors
ax = sns.barplot(x="label", y="episode_return_mean", data=df_sorted, 
                 hue="label", palette='viridis', legend=False)

# Add labels to each bar with formatting
for container in ax.containers:
    ax.bar_label(container, fmt='%.3f', padding=3)

# Set clean axis labels (remove x-axis label)
ax.set_xlabel('')
ax.set_ylabel('Episode Return Mean', fontweight='bold')

# Make tick labels bold
plt.setp(ax.get_xticklabels(), fontweight='bold')
plt.setp(ax.get_yticklabels(), fontweight='bold')

# Add styling
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Create pivot table showing mean episode return by use_residual
pivot = df1.pivot_table(
    values='episode_return_mean',
    index='use_residual',
    aggfunc='mean'
).sort_values('episode_return_mean', ascending=False)

print(pivot)
```

```python
df2 = df.loc[
    (df["exp_id"].isin([17,20, 14]))
]
df2
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a mapping dictionary
period_mapping = {
    0: 'Adaptive Resampling\n(ESS=0.7)',
    4: 'Periodic Resampling\n(period=4)', 
    65: 'No Resampling'
}

# Map the period values to new labels
df2_mapped = df2.copy()
df2_mapped['Resampling Interval'] = df2_mapped['period'].map(period_mapping)

# Find the maximum return for each period and sort by descending order
max_returns = df2_mapped.groupby('Resampling Interval')['episode_return_mean'].max().reset_index()
max_returns_sorted = max_returns.sort_values('episode_return_mean', ascending=False)

# Create the plot with max returns and remove error bars
ax = sns.barplot(x="Resampling Interval", y="episode_return_mean", data=max_returns_sorted, errorbar=None,
                 hue="Resampling Interval", palette='viridis', legend=False)

# Add labels to each bar
for container in ax.containers:
    ax.bar_label(container)

# Make axis labels bold
ax.set_xlabel(ax.get_xlabel(), fontweight='bold')
ax.set_ylabel("Max Episode Return Mean", fontweight='bold')

# Make tick labels bold
plt.setp(ax.get_xticklabels(), fontweight='bold')
plt.setp(ax.get_yticklabels(), fontweight='bold')

plt.show()
```

```python
import pandas as pd

# Create pivot table showing max episode return by resampling strategy
pivot = df2.pivot_table(
    values='episode_return_mean',
    index='period',
    aggfunc='max'
).sort_values('episode_return_mean', ascending=False)

# Map period to readable labels
period_labels = {0: 'ESS=0.7', 4: 'period=4', 65: 'No Resampling'}
pivot.index = pivot.index.map(period_labels)

print(pivot)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Create the raw data
data = {
    'Environment': ['Brax HalfCheetah', 'Brax HalfCheetah', 'Rware', 'Rware', 'DoorKey 16x16', 'DoorKey 16x16'],
    'Resampling Strategy': ['ESS', 'No Resampling', 'ESS', 'No Resampling', 'ESS', 'No Resampling'],
    'Episode Return Mean': [6034, 6245, 3.96, 4.26, 0.91, 0.93]
}

df = pd.DataFrame(data)

# Calculate percentage with No Resampling as 100%
df_pct = df.copy()
for env in df['Environment'].unique():
    no_resample_val = df[(df['Environment'] == env) & (df['Resampling Strategy'] == 'No Resampling')]['Episode Return Mean'].values[0]
    mask = df['Environment'] == env
    df_pct.loc[mask, 'Performance (%)'] = (df.loc[mask, 'Episode Return Mean'] / no_resample_val) * 100

# Create the plot
plt.figure(figsize=(10, 6))
ax = sns.barplot(x='Environment', y='Performance (%)', hue='Resampling Strategy', 
                 data=df_pct, palette=['#ff7f0e', '#2ca02c'])

# Add value labels on bars with % symbol
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', padding=3)

# Make labels bold
ax.set_xlabel('Environment', fontweight='bold')
ax.set_ylabel('Peformance(% relative return)', fontweight='bold')
plt.setp(ax.get_xticklabels(), fontweight='bold')
plt.setp(ax.get_yticklabels(), fontweight='bold')

# Add title and legend
plt.title('ESS vs No Resampling: Relative Performance\n(No Resampling = 100%)', fontweight='bold', pad=20)
plt.legend(title='Resampling Strategy', title_fontproperties={'weight': 'bold'}, 
           loc='upper right', frameon=True)

# Set y-axis to start from a reasonable lower bound to show differences better
plt.ylim(85, 105)

# Add horizontal line at 100%
ax.axhline(y=100, color='gray', linestyle='--', alpha=0.7, linewidth=1)

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Create pivot table showing percentage performance by environment
df_pivot = df_pct.pivot_table(
    values='Performance (%)',
    index='Environment',
    columns='Resampling Strategy',
    aggfunc='mean'
).round(2)

print(df_pivot)
```
