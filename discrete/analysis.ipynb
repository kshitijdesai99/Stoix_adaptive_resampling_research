```python
import pandas as pd
import json

config    = pd.read_csv("config.csv")
absolute  = pd.read_csv("absolute.csv")
evaluator = pd.read_csv("evaluator.csv")
actor     = pd.read_csv("actor.csv")
trainer   = pd.read_csv("trainer.csv")

DEPTH_COLS = ["entropy_depth", "ess_fraction_depth", "mean_td_weights_depth", "particles_alive_depth", "resample_depth"]
for col in DEPTH_COLS:
    trainer[col] = trainer[col].apply(lambda x: json.loads(x) if pd.notna(x) and x != "" else [])

df = (
    config
    .merge(absolute,  on="exp_id", suffixes=("", "_abs"))
    .merge(evaluator, on="exp_id", suffixes=("", "_eval"))
    .merge(actor,     on="exp_id", suffixes=("", "_actor"))
    .merge(trainer,   on="exp_id", suffixes=("", "_trainer"))
)

df.shape
```

```python
df.to_csv("analysis.csv", index=False)
```
